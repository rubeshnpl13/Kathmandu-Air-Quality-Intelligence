## Project overview

This project analyzes hourly air-pollution and meteorological data for Kathmandu from **September 2022 through April 2025**.

The complete workflow covers:

- data loading and integration
- data-quality assessment
- feature engineering
- particulate AQI calculation
- exploratory data analysis
- time-series diagnostics
- statistical and meteorological analysis
- extreme-pollution event analysis
- leakage-safe ML forecasting
- model comparison
- explainability and forecast-error analysis
- interactive visualization through Streamlit

### Research questions

1. What temporal patterns exist in Kathmandu air pollution?
2. Which meteorological conditions are associated with pollution?
3. How persistent is PM2.5 over hourly, daily, and weekly timescales?
4. What conditions characterize extreme pollution episodes?
5. Can PM2.5 be forecast 24 hours ahead better than a strong persistence baseline?

## Key findings

### Temporal and seasonal structure
- Winter had the highest PM2.5 concentrations.
- Monsoon had the lowest concentrations.
- PM2.5 displayed a pronounced diurnal cycle.
- Weekday/weekend differences were small compared with seasonal and hourly structure.

### Persistence
Approximate hourly PM2.5 lag correlations included:

- **1 hour:** 0.965
- **24 hours:** 0.877
- **168 hours:** 0.687

The strong 24-hour persistence makes a same-hour-previous-day forecast a difficult benchmark to beat.

### Extreme pollution
- PM2.5 90th percentile: **66.1**
- PM2.5 95th percentile: **83.6**
- Top-10% threshold produced **303 episodes**
- **254** episodes lasted at least 2 hours
- Median duration: **6 hours**
- Longest episode: **65 hours**

### Forecasting
Held-out test performance:

| Model | MAE | RMSE | R² |
|---|---:|---:|---:|
| Persistence | 12.085 | 16.284 | 0.738 |
| **Ridge** | **11.314** | **15.028** | **0.777** |

The frozen Ridge model reduced test MAE by approximately **6.4%** relative to persistence.

### Explainability
- Recent pollution dynamics were the strongest predictive inputs.
- Ridge added the most value when PM2.5 changed substantially over 24 hours.
- Persistence remained extremely competitive under stable conditions.
- Ridge tended to underpredict the highest observed PM2.5 concentrations.

## Important reproducibility notes

- Run the notebooks in numerical order for full reproduction.
- Keep the repository structure unchanged because phases exchange saved CSV outputs.
- The forecasting workflow uses chronological splits rather than random train/test splitting.
- Learned preprocessing must be fitted only on training/development observations.
- The held-out test period must not be used for model or hyperparameter selection.
- Phase 10 is post-hoc diagnostics and should not be used to retune the model.

### Scientific limitations

- Pollutant unit symbols were mangled in the original exported headers, so exact pollutant units should be verified from source metadata before formal publication.
- Processed timestamps are timezone-naive; hour-of-day conclusions use timestamps as supplied.
- The AQI analysis is a calculated **particulate AQI** based on PM2.5 and PM10, not a claimed full official multi-pollutant Kathmandu AQI.
- Meteorological relationships are observational associations and should not be interpreted as causal effects.
