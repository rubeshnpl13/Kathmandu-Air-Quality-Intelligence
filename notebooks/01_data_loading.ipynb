```python
#missing value analysis
def missing_summary(df):
    summary = pd.DataFrame({
        'missing_count': df.isna().sum(),
        'missing_percent': df.isna().mean() * 100
    })

    return summary.sort_values(
        by='missing_percent',
        ascending=False
    )
```

```python
#missing timestamps
expected_pollution_index = pd.date_range(
    start=pollution['time'].min(),
    end=pollution['time'].max(),
    freq='h'
)

missing_pollution_times = (
    expected_pollution_index
    .difference(
        pd.DatetimeIndex(pollution['time'])
    )
)

print(
    "Expected pollution observations:",
    len(expected_pollution_index)
)

print(
    "Actual pollution observations:",
    len(pollution)
)

print(
    "Missing pollution timestamps:",
    len(missing_pollution_times)
)
```

```python
expected_weather_index = pd.date_range(
    start=weather['time'].min(),
    end=weather['time'].max(),
    freq='h'
)

missing_weather_times = (
    expected_weather_index
    .difference(
        pd.DatetimeIndex(weather['time'])
    )
)

print(
    "Expected weather observations:",
    len(expected_weather_index)
)

print(
    "Actual weather observations:",
    len(weather)
)

print(
    "Missing weather timestamps:",
    len(missing_weather_times)
)
```

```python
#temporal completeness
pollution_completeness = (
    len(pollution)
    / len(expected_pollution_index)
    * 100
)

weather_completeness = (
    len(weather)
    / len(expected_weather_index)
    * 100
)

print(
    f"Pollution temporal completeness: "
    f"{pollution_completeness:.2f}%"
)

print(
    f"Weather temporal completeness: "
    f"{weather_completeness:.2f}%"
)
```

```python
#time overlap before merging
overlap_start = max(
    pollution['time'].min(),
    weather['time'].min()
)

overlap_end = min(
    pollution['time'].max(),
    weather['time'].max()
)

print("Common start:", overlap_start)
print("Common end:", overlap_end)
```

```python
pollution_only_times = (
    set(pollution['time'])
    - set(weather['time'])
)

weather_only_times = (
    set(weather['time'])
    - set(pollution['time'])
)

print(
    "Pollution timestamps without weather:",
    len(pollution_only_times)
)

print(
    "Weather timestamps without pollution:",
    len(weather_only_times)
)
```

```python
#merging dataset
df = pd.merge(
    pollution,
    weather,
    on='time',
    how='inner',
    validate='one_to_one'
)
```

## Summary: Data Acquisition and Integration

The air-pollution and meteorological datasets were imported separately,
their timestamps were standardized, and both datasets were evaluated for
missing records, duplicate timestamps, invalid values, and temporal
continuity.

The datasets were then aligned using their hourly timestamps through a
one-to-one inner join. The resulting integrated dataset contains pollution
measurements (PM2.5, PM10, NO₂, SO₂, and CO) together with meteorological
variables including temperature, rainfall, wind direction, wind speed,
relative humidity, surface pressure, and dew point.

The merged dataset was sorted chronologically and stored using a
DatetimeIndex to support subsequent time-series analysis.

No outlier removal or statistical transformations were performed during
this stage. Data-quality issues and potential extreme observations will be
examined separately during Phase 2.
